```python
import geopy
from geopy.geocoders import Nominatim
from geopy.extra.rate_limiter import RateLimiter

geolocator = Nominatim(user_agent="jam")
# Taken from documentation: Retrieves the standardized geocode
# geocode_limited = RateLimiter(geolocator.geocode, min_delay_seconds=1)
```

```python
location = geolocator.geocode(address)
location.address
```

```python
home = geolocator.geocode("Samsøgade 50 9000 Aalborg")
home.address
```

```python
from geopy.distance import geodesic

dist = geodesic(
    (home.latitude, home.longitude), (location.latitude, location.longitude)
).kilometers
dist
```
