```python
import time

from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By
```

```python
# Path to your chromedriver
service = Service("/path/to/chromedriver")
driver = webdriver.Chrome(service=service)

driver.get(
    "https://www.boligsiden.dk/adresse/gerdingvej-35-9520-skoerping-08400265__35_______?udbud=c61ad15c-c0fc-4bf7-afbe-a1a5d886f43d"
)

# Wait for page to load (optional: use WebDriverWait for dynamic content)
time.sleep(3)

# Example: get page source and parse with BeautifulSoup
soup = BeautifulSoup(driver.page_source, "html.parser")

# Now you can extract elements as usual
print(soup.find("h1").text)

driver.quit()
```
