```python
gmaps = googlemaps.Client(key=api)
origin = "Gerdingvej 35, 9520 Skørping, Denmark"
destinations = [
    "Søndervang 4, 7870 Roslev, Denmark",
    "Søndertoften 24, 9500 Hobro, Denmark",
    "Samsøgade 50, 9000 Aalborg, Denmark",
]
_destinations = {
    "durup": "Søndervang 4, 7870 Roslev, Denmark",
    "hobro": "Søndertoften 24, 9500 Hobro, Denmark",
    "net": "Gasværksvej 20, 9000 Aalborg, Denmark",
    "wrist": "Stigsborgvej 60, 9400 NørresundbyS, Denmark",
}
```

```python
list(_destinations.values())
```

```python
result = gmaps.distance_matrix(
    origins=[origin], destinations=list(_destinations.values()), mode="driving"
)
```

```python
for i, element in enumerate(result["rows"][0]["elements"]):
    print(f"Travel time to {destinations[i]}: {element['duration']['text']}")
```

```python
distances = {}

for i, element in enumerate(result["rows"][0]["elements"]):
    distances[destinations[i]] = element["duration"]["text"]

for k, v in distances.items():
    print(f"{k}: {v}")
```
